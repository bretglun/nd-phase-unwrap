test if masking does a difference to unwrapping performance
test 3+1D phantom with VNR 3.0

Unwrap the rawfile without mask and compare with the output from the masked run already available in IGC_unwrapped.
The comparison shall only be done inside the mask, meaning inside the volume and all the background shall not be considered.

```python
from pathlib import Path
import sys
REPO_ROOT = Path.cwd().parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
from nd_phase_unwrap import io
import os 
import numpy as np

data_path = REPO_ROOT / 'test_data'
in_path = data_path / '3+1D_phantom'
out_path_mask = data_path / '4DphantomMask'
out_path_nomask = data_path / '4DphantomNoMask'

mask = np.angle(io.load_numpy_data(in_path/"0wraps.npy")) != 0
for file in [j for j in os.listdir(in_path) if (j.endswith('1_0.npy') or j.endswith('3_0.npy'))]:
    print (f'Processing file {file}...')
    file = Path(file)
    data_file_mask = out_path_mask / f'{file.stem}_unwrapped.npy'
    data_file_nomask = out_path_nomask / f'{file.stem}_unwrapped.npy'
    
    array_mask = io.load_numpy_data(data_file_mask)
    array_nomask = io.load_numpy_data(data_file_nomask)

    diff = np.abs(array_mask - array_nomask)


    if np.max(diff[mask]) != 0:
        print(f'{file} diff bigger than 0! {np.sum(diff[mask] != 0)} voxels and max diff is {np.max(diff[mask])}')
```
